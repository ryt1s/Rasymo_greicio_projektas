```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/rasymo_sesija.csv")
```

```python
avg_progress = df.groupby("sesijos_id")["wpm"].mean()

plt.figure()
plt.plot(avg_progress.index, avg_progress.values)
plt.title("Learning Curve (WPM progresas per sesijas)")
plt.xlabel("Sesija")
plt.ylabel("Vidutinis WPM")
plt.show()
```

```python
user_avg = df.groupby("vartotojo_id")["wpm"].mean()

plt.figure()
plt.hist(user_avg, bins=20)
plt.title("Vartotojų vidutinis WPM pasiskirstymas")
plt.xlabel("WPM")
plt.ylabel("Kiek vartotojų")
plt.show()
```
