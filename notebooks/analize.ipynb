```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/rasymo_sesija.csv")

df.head()
```

```python
avg_wpm = df.groupby("sesijos_id")["wpm"].mean()

plt.figure()
plt.plot(avg_wpm.index, avg_wpm.values)
plt.title("Vidutinis WPM laikui bėgant")
plt.xlabel("Sesija")
plt.ylabel("WPM")
plt.show()
```

```python
plt.figure()
plt.scatter(df["wpm"], df["tikslumas"], alpha=0.5)
plt.title("WPM vs Tikslumas")
plt.xlabel("WPM")
plt.ylabel("Tikslumas (%)")
plt.show()
```

```python
for user_id in df["vartotojo_id"].unique()[:5]:
    user_data = df[df["vartotojo_id"] == user_id]
    plt.plot(user_data["sesijos_id"], user_data["wpm"], alpha=0.6)

plt.title("Skirtingų vartotojų WPM progresas")
plt.xlabel("Sesija")
plt.ylabel("WPM")
plt.show()
```
